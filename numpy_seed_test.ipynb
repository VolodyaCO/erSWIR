```python
from graph_tool.all import *
import numpy as np
from pathos.multiprocessing import ProcessingPool as Pool
import tqdm
```

```python
def swir(n, z, num_ensamble):
    pER = z/n
    np.random.seed(num_ensamble)
    seed_rng(num_ensamble)
    # inicialización del grafo erdos renyi
    g = random_graph(n, lambda: np.random.poisson((n-1) * pER), directed=False, model="erdos", random=True)
    return g
```

```python
def worker_function(num_ensamble):
    return swir(10000, 8, num_ensamble)
```

```python
ensambles = 4
with Pool(4) as pool:
    results = list(tqdm.tqdm(pool.imap(worker_function, range(ensambles)), total=ensambles))
```
