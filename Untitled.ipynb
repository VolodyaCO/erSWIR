```python
import numpy as np
from scipy.optimize import fsolve, newton
```

```python
def mp(kappa, rho):
    a = kappa*8 - 1/(1-rho)
    b = 0.5*kappa*64*(0.5-2*kappa)
    c = (1/6)*kappa*(4*kappa*kappa - kappa - 0.25)*8*8*8
    return -b/(3*c) + np.sqrt(b*b/(9*c*c)-a/(3*c))

def mm(kappa, rho):
    a = kappa*8 - 1/(1-rho)
    b = 0.5*kappa*64*(0.5-2*kappa)
    c = (1/6)*kappa*(4*kappa*kappa - kappa - 0.25)*8*8*8
    return -b/(3*c) - np.sqrt(b*b/(9*c*c)-a/(3*c))
```

```python
def F(m, kappa, rho):
    a = kappa*8 - 1/(1-rho)
    b = 0.5*kappa*64*(0.5-2*kappa)
    c = (1/6)*kappa*(4*kappa*kappa - kappa - 0.25)*8*8*8
    return rho/(1-rho) + a*m + b*m*m + c*m*m*m
```

```python
def find_rho(rho):
    fp = lambda kappa: F(mp(kappa, rho), kappa, rho)
    fm = lambda kappa: F(mm(kappa, rho), kappa, rho)
    kappap = newton(fp, 0.108)
    kappam = newton(fm, 0.108)
    return kappap[0] - kappam[0]
```

```python
fsolve(find_rho, 0.00747)
```
