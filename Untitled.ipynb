```python
def mm(z, rho, eta):
    ka = 1/(z*(1-rho))
    b = 0.5*ka*(eta-2*ka)*z*z
    c = (1/6)*ka*(4*ka*ka-2*eta*ka-eta*eta)*z*z*z
    return -(2/3)*b/c
```

```python
mm(10, 0.1, 0.2)
```

```python
mm(8, 2e-3, 0.5)
```

```python
def mm(y):
    return (y-2)/(y*y-y+4)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ys = np.linspace(0, 4, 100)
mms = mm(ys)
```

```python
plt.rc('text', usetex=True)
plt.plot(ys, mms, 'k-')
plt.axhline(0, c='r', ls='--')
plt.axvline(2, c='r', ls='--')
plt.xlabel(r'$y$', fontsize=14)
plt.ylabel(r'$m^-$', fontsize=14)
plt.yticks([0], ['0'])
plt.savefig('kappa_a_wrong.pdf', dpi=200)
plt.show()
```
