```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
from scipy.optimize import minimize
```

```python
eta = 0.5
z = 8
```

```python
def f(m, kappa):
    return 1 - (1+kappa/(eta-2*kappa))*np.exp(-2*kappa*m*z) + kappa/(eta-2*kappa)*np.exp(-eta*m*z)
```

```python
def F(m,kappa, rho0):
    return (1-rho0)*f(m,kappa) - (m-rho0)
```

```python
rho0 = [0.00747762]
for j in range(0,len(rho0)):
    kappa = [0.1101,0.108021]
    for i in range(0,len(kappa)):
        m = np.arange(0.0,0.4,0.002)
        plt.axhline(0,0,1)
        plt.plot(m,F(m,kappa[i], rho0[j]))
    plt.show()
#plt.xlabel(r'$m$', fontsize = 16) 
#plt.ylabel(r'$F(m,\rho_0)$',fontsize = 16) 
```
