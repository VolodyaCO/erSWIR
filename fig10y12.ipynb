```python
import os
import pickle
import numpy as np
import matplotlib.pyplot as plt
```

```python
N1 = 1000000
kappa_range = (np.linspace(-1, 1, 20))**3
max_new = 0.115+0.003
min_new = 0.115-0.003
kappa_range = (kappa_range-kappa_range.min()) * (max_new - min_new) / 2 + min_new

num_ensambles = 3000
M1 = np.empty((len(kappa_range), num_ensambles))
for n in range(num_ensambles):
    with open('results/{0}_{1}_rho_critical.p'.format(n, N1), "rb") as f:
        mags = pickle.load(f)
        for k in range(len(kappa_range)):
            m = mags[k]
            M1[k,n] = m
```

```python
N2 = 50000
kappa_range = (np.linspace(-1, 1, 20))**3
max_new = 0.115+0.003
min_new = 0.115-0.003
kappa_range = (kappa_range-kappa_range.min()) * (max_new - min_new) / 2 + min_new

num_ensambles = 20000
M2 = np.empty((len(kappa_range), num_ensambles))
for n in range(num_ensambles):
    with open('results/{0}_{1}_rho_critical.p'.format(n, N2), "rb") as f:
        mags = pickle.load(f)
        for k in range(len(kappa_range)):
            m = mags[k]
            M2[k,n] = m
```

```python
m0_1M = np.mean(M1[10][M1[10]<0.15])
m0_50k = np.mean(M2[10][M2[10]<0.15])
m0_100k = np.mean(M3[M3 < 0.15])
m0_146k = np.mean(M4[M4 < 0.15])
```
