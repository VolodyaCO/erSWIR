```python
import pickle
import os
import numpy as np
```

```python
for filename in os.listdir('results'):
    print(filename)
    with open("results/{0}".format(filename), "rb") as f:
        content = pickle.load(f)
    content = np.array(content)
    with open("results/{0}".format(filename), "wb") as f:
        pickle.dump(content, f)
```
