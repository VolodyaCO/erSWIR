```python
import numpy as np
from scipy.optimize import fsolve
```

```python
def F(m, *args):
    rho0, kappa, z, eta = args
    a = kappa*z - 1/(1-rho0)
    b = 0.5*kappa*z*z*(eta - 2*kappa)
    c = (1/6) * kappa*z*z*z*(4*kappa*kappa - 2*eta*kappa - eta*eta)
    return rho0/(1-rho0) + a*m + b*m*m + c*m*m*m
```

```python
Fs = [F(m, (2e-3, 0.7, 8, 0.5)) for m in np.linspace(0, 1, 1000)]
m = np.array([0, 1])
fsolve(F, m, args = (2e-3, 0.7, 8, 0.5))
```
